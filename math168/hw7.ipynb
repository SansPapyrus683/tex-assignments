# HW 7 Degree Simulations

```python
import numpy as np
import networkx as nx
```

## Part A

```python
def degree_distr(lam: int, sz: int):
    deg = np.random.poisson(lam=lam, size=sz)
    if sum(deg) % 2 == 1:
        deg[0] += 1
    return deg
```

```python
n = 1000
deg = degree_distr(12, n)

avg_k_sq = (deg ** 2).mean()
avg_k = deg.mean()

m_wrong = 1/2 * ((avg_k_sq - avg_k) / avg_k) ** 2
m_correct = m_wrong / 2

print(f"M_wrong: {m_wrong}")
print(f"M_correct: {m_correct}")
```

## Part B

```python
samples = []
for _ in range(100):
    g = nx.configuration_model(deg)

    num_multiedges = 0
    for i in range(n):
        for j in range(i + 1, n):
            num_multiedges += max(0, g.number_of_edges(i, j) - 1)
    samples.append(num_multiedges)

print(f"Empirical M: {np.mean(samples)}")
```
