# HW 8 Wattz-Strogatz Graphs

## Part A

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

```python
n = 1000
k = 6

to_test = [1e-5, 5e-5, 1e-4, 5e-4, 1e-3, 5e-3, 1e-2, 5e-2, .1, .2, .3]

all_p = []
all_diameters = []
avg_diameters = []
for p in to_test:
    curr_diameters = []
    for _ in range(25):
        g = nx.watts_strogatz_graph(n, k, p)
        d = nx.diameter(g)

        all_p.append(p)
        all_diameters.append(d)
        curr_diameters.append(d)

    avg_diameters.append(np.mean(curr_diameters))
```

```python
plt.scatter(all_p, all_diameters, marker=".", color="black")
plt.plot(to_test, avg_diameters, marker="o")
plt.xlabel("p")
plt.xscale("log")
plt.ylabel("Diameter")
```

## Part B

Intuitively, the WS model seems like a pretty good approximation of social networks.
Each person probably knows most of the people in their general area, save for
a couple friends or family members that might be across the globe.
It does assume that everyone has the same number of friends, though, which is just not true.
Another assumption it makes is that the distribution for far-away friends is uniformly
random, when it could be dependent on some external factors.
For example, an Asian American would be more likely to know someone in far-away China than in the UK.
